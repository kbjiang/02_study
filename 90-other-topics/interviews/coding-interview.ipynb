```python
# 209, Minimum size subarray sum
def minSubArrayLen(target: int, nums):
    r = 0
    curList = []

    minLen = len(nums)
    # Had I used slicing, the right boundary condition would be weird.
    # for r in range(len(nums)+1):
    #     curList= nums[l:r]
    for r in range(len(nums)): 
        curList.append(nums[r])
        while sum(curList) >= target:
            minLen = min(minLen, len(curList))
            curList = curList[1:]
            # print(curList)
    if minLen == len(nums) and sum(nums) < target:
        return 0
    return minLen
```

```python
# 76. Minimum Window Substring
def minWindow(s, t):
    l = 0
    r = 0
    minWin = "x"*(max(len(s), len(t))+1)
    t_counts = {char: t.count(char) for char in set(t)}

    for r in range(len(s)):
        if s[r] in t_counts:
            t_counts[s[r]] -= 1
        while all([cnt <= 0 for cnt in t_counts.values()]):
            minWin_ = s[l:r+1]
            minWin = min(minWin, minWin_, key=len)
            if s[l] in t_counts:
                t_counts[s[l]] += 1
            l += 1
    if minWin == "x"*(max(len(s), len(t))+1):
        return ""
    return minWin
```

```python
# 121. Best time to buy and sell stock
def maxProfit(prices:int) -> int:
    if len(prices) <= 1:
        return 0
    l = 0
    max_profit = 0
    for r in range(1, len(prices)):
        profit = prices[r] - prices[l]
        if profit < 0:
            l = r
            continue
        elif profit > max_profit:
            print(l, r)
            max_profit = profit
    # boundary condition
    if l == len(prices) - 1:
        return 0
    return max_profit
```

## Hashtable
**Use *Two Sum* as an example to show procedure**
1. "exactly one solution" means can stop early and no worries about running outta boundary
1. no "same element twice" but different elements can have same value
1. edge case:
    1. diff elem with same value can lead to collison, so we should search before insert

```python
# 1. Two sum
def twoSum(nums, target):
    tbl = dict()
    for i, num in enumerate(nums):
        candidate = target - num
        if candidate in tbl:  # O(1)
            return tbl[candidate], i
        tbl[num] = i
```

```python
# 509 Fibonacci number (with memoization)
def fib(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib(n-1, memo) + fib(n-2, memo)
    return memo[n]
```

22, Generate Parentheses
- base case happens at max length, i.e., only valid full length strings make it to the result list.
- recursion conditions, `n_left < n` select valid paths

```python
# 22 Generate Parentheses
def generateParenthesis(self, n: int) -> list[str]:
	res = []
	def backtrack(n_left, n_right, s) -> None:
		# n_left: num of left parenthesis
		# n_right: num of right parenthesis
		
		# base case
		if len(s) == n*2:
			res.append(s)
			# return 
		
		# either `if+return` or `if+else``
		else:
			# attempt to add left parenthesis
			if n_left < n:
				backtrack(n_left + 1, n_right, s+"(")
			# attempt to add right parenthesis
			if n_right < n_left: # num of rights never exceeds num of lefts
				backtrack(n_left, n_right+1, s+")")

	backtrack(0, 0, "")
	return res
```

77, Combinations
- the explicit backtrack of `lst`
- use of `lst.copy()` when appending

```python
# 77 Combinations
def combine(self, n: int, k: int) -> list[list[int]]:
	res = []
	def backtrack(lst, cans):
		# base case
		if len(lst) == k:		
		# .copy to not affect items in `res`
		res.append(lst.copy())
		return
		
		# attempt possible actions
		for i, can in enumerate(cans):
			lst = lst + [can]
			backtrack(lst, cans[i+1:])
			# to backtrack
			lst.pop()

	backtrack([], list(range(1, n+1)))
	return res
```
