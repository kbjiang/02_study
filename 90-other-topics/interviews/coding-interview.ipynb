```python
# 76. Minimum Window Substring
def minWindow(s, t):
    l = 0
    r = 0
    minWin = "x"*(max(len(s), len(t))+1)
    t_counts = {char: t.count(char) for char in set(t)}

    for r in range(len(s)):
        if s[r] in t_counts:
            t_counts[s[r]] -= 1
        while all([cnt <= 0 for cnt in t_counts.values()]):
            minWin_ = s[l:r+1]
            minWin = min(minWin, minWin_, key=len)
            if s[l] in t_counts:
                t_counts[s[l]] += 1
            l += 1
    if minWin == "x"*(max(len(s), len(t))+1):
        return ""
    return minWin
```

```python
# 1. Two sum
def twoSum(nums, target):
    tbl = dict()
    for i, num in enumerate(nums):
        candidate = target - num
        if candidate in tbl:  # O(1)
            return tbl[candidate], i
        tbl[num] = i
```

```python
# 509 Fibonacci number (with memoization)
def fib(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib(n-1, memo) + fib(n-2, memo)
    return memo[n]
```

22, Generate Parentheses
- base case happens at max length, i.e., only valid full length strings make it to the result list.
- recursion conditions, `n_left < n` select valid paths

```python
# 22 Generate Parentheses
def generateParenthesis(self, n: int) -> list[str]:
	res = []
	def backtrack(n_left, n_right, s) -> None:
		# n_left: num of left parenthesis
		# n_right: num of right parenthesis
		
		# base case
		if len(s) == n*2:
			res.append(s)
			# return 
		
		# either `if+return` or `if+else``
		else:
			# attempt to add left parenthesis
			if n_left < n:
				backtrack(n_left + 1, n_right, s+"(")
			# attempt to add right parenthesis
			if n_right < n_left: # num of rights never exceeds num of lefts
				backtrack(n_left, n_right+1, s+")")

	backtrack(0, 0, "")
	return res
```

77, Combinations
- the explicit backtrack of `lst`
- use of `lst.copy()` when appending

```python
# 77 Combinations
def combine(self, n: int, k: int) -> list[list[int]]:
	res = []
	def backtrack(lst, cans):
		# base case
		if len(lst) == k:		
		# .copy to not affect items in `res`
		res.append(lst.copy())
		return
		
		# attempt possible actions
		for i, can in enumerate(cans):
			lst = lst + [can]
			backtrack(lst, cans[i+1:])
			# to backtrack
			lst.pop()

	backtrack([], list(range(1, n+1)))
	return res
```

```python
def maxDepth(root: TreeNode | None) -> int:
    max_depth = 0

    def recur(u, depth):
        if not u:
            max_depth = max(max_depth, depth)
            return
        else:
            recur(u.left, depth+1)
            recur(u.right, depth+1)

    recur(root, 0)
    return max_depth
```

```python
def graph_topo_sort(num_nodes, edges):
    from collections import deque
    nodes, order, queue = {}, [], deque()
    for node_id in range(num_nodes):
        nodes[node_id] = { 'in': 0, 'out': set() }
    for node_id, pre_id in edges:
        nodes[node_id]['in'] += 1
        nodes[pre_id]['out'].add(node_id)
    for node_id in nodes.keys():
        if nodes[node_id]['in'] == 0:
            queue.append(node_id)
    while len(queue):
        node_id = queue.popleft()
        for outgoing_id in nodes[node_id]['out']:
            nodes[outgoing_id]['in'] -= 1
            if nodes[outgoing_id]['in'] == 0:
                queue.append(outgoing_id)
        order.append(node_id)
    return order if len(order) == num_nodes else None

print(graph_topo_sort(4, [[0, 1], [0, 2], [2, 1], [3, 0]]))
# [1, 2, 0, 3]
```

```python
def dfs(matrix):
  # Check for an empty matrix/graph.
  if not matrix:
    return []

  rows, cols = len(matrix), len(matrix[0])
  visited = set()
  directions = ((0, 1), (0, -1), (1, 0), (-1, 0))

  def traverse(i, j):
    if (i, j) in visited:
      return

    visited.add((i, j))
    # Traverse neighbors.
    for direction in directions:
      next_i, next_j = i + direction[0], j + direction[1]
      if 0 <= next_i < rows and 0 <= next_j < cols:
        # Add in question-specific checks, where relevant.
        traverse(next_i, next_j)

  for i in range(rows):
    for j in range(cols):
      traverse(i, j)
```

```python
def lengthOfLongestSubstring(s):
    if len(s) == 0:
        return 0

    i = 0
    j = 1
    maxLen = 1

    curChars = set(s[i])
    while j < len(s):
        if s[j] not in curChars:
            curChars.add(s[j])
            # print(curChars)
            j += 1
            maxLen = max(maxLen, len(curChars))
        else:
            while i < j:
                curChars.remove(s[i])
                i += 1
            curChars.add(s[j])
            j += 1

    return maxLen
```

```python
def permute(nums: list[int]):
    res = []

    def backtrack(perm, start):
        if start == len(nums):
            res.append(perm.copy())
            return              
        
        for i in range(len(nums)):
            perm.append(nums[i])
            backtrack(perm, i+1)
            perm.pop()

    backtrack([], 0)
    return res
```

```python
def letterCombinations(digits: str) -> list[str]:
    digit2char = {
       "2": ['a', 'b', 'c'],
       "3": ['d', 'e', 'f'],
       "4": ['g', 'h', 'i'],
       "5": ['j', 'k', 'l'],
       "6": ['m', 'n', 'o'],
       "7": ['p', 'q', 'r', 's'],
       "8": ['t', 'u', 'v'],
       "9": ['w', 'x', 'y', 'z'],
    }

    combs = digit2char[digits[0]]

    for digit in digits[1:]:
        combs_ = []
        for char in digit2char[digit]:
            combs_.extend([comb + char for comb in combs])
        combs = combs_.copy()

    return combs
```

```python
def kthSmallest(matrix: list[list[int]], k: int) -> int:
    import math
    n = math.ceil(math.sqrt(k))
    print(n)

    if n**2 == k:
        return matrix[n-1][n-1]

    row = 0
    col = 0
    counts = k - (n-1)**2
    print(counts)

    while counts > 0:
        while row <= n-1 and col <= n-1:
            if matrix[n-1][col] <= matrix[row][n-1]:
                col += 1
            else:
                row += 1
            counts -= 1
    if col == n:
        return matrix[row+counts][n-1]
    else:
        return matrix[n-1][col+counts]
```
